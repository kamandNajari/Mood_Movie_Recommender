```python
import cv2
from deepface import DeepFace
import random

emotion_to_genres = {
    "happy": ["Comedy", "Adventure", "Animation"],
    "sad": ["Comedy", "Feel-good Drama"],
    "angry": ["Action", "Thriller"],
    "surprise": ["Mystery", "Sci-Fi"],
    "fear": ["Comedy", "Family"],
    "disgust": ["Dark Comedy", "Satire"],
    "neutral": ["Drama", "Documentary"]
}

movie_database = {
    "Comedy": ["Superbad", "The Grand Budapest Hotel", "Groundhog Day", "Bridesmaids"],
    "Adventure": ["Raiders of the Lost Ark", "The Goonies", "Jumanji"],
    "Animation": ["Spirited Away", "Up", "Coco", "Inside Out"],
    "Feel-good Drama": ["The Pursuit of Happyness", "Forrest Gump", "Life Is Beautiful"],
    "Action": ["Mad Max: Fury Road", "John Wick", "Die Hard"],
    "Thriller": ["Se7en", "Gone Girl", "Prisoners"],
    "Mystery": ["Knives Out", "Shutter Island", "The Prestige"],
    "Sci-Fi": ["Interstellar", "Arrival", "Inception"],
    "Family": ["The Incredibles", "Paddington", "E.T. the Extra-Terrestrial"],
    "Dark Comedy": ["In Bruges", "The Death of Stalin", "Fargo"],
    "Satire": ["Don't Look Up", "Get Out", "American Psycho"],
    "Drama": ["The Shawshank Redemption", "Good Will Hunting", "Parasite"],
    "Documentary": ["Free Solo", "My Octopus Teacher", "13th"]
}

cap = cv2.VideoCapture(0)
print("Press SPACE to capture and analyze your expression. Press q to quit.")

captured_frame = None

while True:
    ret, frame = cap.read()
    if not ret:
        break
    frame = cv2.flip(frame, 1)
    display = frame.copy()
    cv2.putText(display, "Press SPACE to capture", (20, 40),
                cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2)
    cv2.imshow("Mood Movie Recommender", display)

    key = cv2.waitKey(1) & 0xFF
    if key == 32:
        captured_frame = frame.copy()
        break
    elif key == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()

if captured_frame is not None:
    result = DeepFace.analyze(
        captured_frame,
        actions=["emotion"],
        enforce_detection=False,
        detector_backend="skip"
    )
    dominant_emotion = result[0]["dominant_emotion"]
    print(f"Detected emotion: {dominant_emotion}")

    genres = emotion_to_genres.get(dominant_emotion, ["Drama"])
    chosen_genre = random.choice(genres)
    recommendations = random.sample(movie_database[chosen_genre], min(3, len(movie_database[chosen_genre])))

    print(f"Suggested genre: {chosen_genre}")
    print("Recommended movies:")
    for movie in recommendations:
        print(f" - {movie}")
else:
    print("No image captured.")
```
